```python
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(numbers))
print("Mean:", np.mean(numbers))
print("Maximum:", np.max(numbers))
print("Minimum:", np.min(numbers))
```

```python
numbers = np.array([10, 20, 30, 40, 50])

print(numbers[0])
print(numbers[2])
print(numbers[-1])
```

```python
numbers = np.array([10, 20, 30, 40, 50])

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
```

```python
marks = np.array([
    [80, 75, 90],
    [65, 70, 85],
    [90, 88, 95]
])

print(marks)
print("Shape:", marks.shape)
```

```python
print(marks[0, 0])
print(marks[1, 2])
print(marks[2, 1])
```

```python
marks = np.array([
    [80, 75, 90],
    [65, 70, 85],
    [90, 88, 95]
])

print("Overall Average:", np.mean(marks))
print("Highest Mark:", np.max(marks))
print("Lowest Mark:", np.min(marks))
```
