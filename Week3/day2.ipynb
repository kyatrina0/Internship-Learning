```python
import pandas as pd
data = {
    "Name": ["Alice", "Badal", "Chie", "David", "Emma"],
    "Department": ["HR", "IT", "IT", "Finance", "HR"],
    "Salary": [40000, 60000, 55000, 50000, 45000]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.head())
print(df.tail())
print(df.shape)
print(df.columns)
print(df.info())
```

```python
print(df["Name"])
print(df["Salary"])
```

```python
print(df[df["Salary"] > 50000])
print(df[df["Department"] == "IT"])
```

```python
print(df["Salary"].mean())
print(df["Salary"].max())
print(df["Salary"].min())
print(df["Salary"].sum())
```
