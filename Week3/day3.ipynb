```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma", "Alice"],
    "Age": [23, 25, None, 30, 28, 23],
    "Salary": [40000, 60000, 55000, None, 45000, 40000],
    "Department": ["HR", "IT", "IT", "Finance", "HR", "HR"]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.isnull())
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df = df.drop_duplicates()
print(df)
print(df.isnull().sum())
```

```python
df.to_csv("cleaned_employees.csv", index=False)
```
