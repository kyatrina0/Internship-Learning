```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Department": ["HR", "IT", "IT", "Finance", "HR"],
    "Salary": [40000, 60000, 55000, 50000, 45000],
    "Experience": [2, 5, 4, 6, 3]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Annual Salary"] = df["Salary"]
df["Monthly Salary"] = df["Salary"] / 12
df["Experience Level"] = df["Experience"].apply(
    lambda x: "Experienced" if x >= 5 else "Beginner"
)

print(df)
```

```python
df = df.sort_values("Salary", ascending=False)

print(df)
```

```python
df = df.rename(columns={
    "Salary": "Annual Salary"
})

print(df)
```

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Department": ["HR", "IT", "IT", "Finance", "HR"],
    "Salary": [40000, 60000, 55000, 50000, 45000],
    "Experience": [2, 5, 4, 6, 3]
}

df = pd.DataFrame(data)

# Create monthly salary
df["Monthly Salary"] = df["Salary"] / 12

# Create experience level
df["Experience Level"] = df["Experience"].apply(
    lambda x: "Experienced" if x >= 5 else "Beginner"
)

# Display employees earning above 50,000
print(df.loc[df["Salary"] > 50000])

# Display first 3 rows
print(df.iloc[0:3])
```
