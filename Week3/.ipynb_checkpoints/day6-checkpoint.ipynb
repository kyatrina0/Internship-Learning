```python
import pandas as pd

data = {
    "Name": [
        "Alice", "Bob", "Charlie", "David",
        "Emma", "Frank", "Grace", "Henry"
    ],
    "Department": [
        "HR", "IT", "IT", "Finance",
        "HR", "IT", "Finance", "HR"
    ],
    "Age": [23, 28, 25, 32, 27, 30, 35, 26],
    "Salary": [
        40000, 60000, 55000, 50000,
        45000, 65000, 70000, 42000
    ],
    "Experience": [1, 5, 3, 7, 4, 6, 10, 2]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("Total employees:", len(df))
```

```python
print("Average salary:", df["Salary"].mean())
```

```python
print("Highest salary:", df["Salary"].max())
```

```python
print(
    df.loc[df["Salary"].idxmax()]
)
```

```python
print(
    df.groupby("Department")["Salary"].mean()
)
```

```python
print(
    df["Department"].value_counts()
)
```

```python
print(
    df[df["Salary"] > 50000]
)
```

```python
print(
    df[df["Experience"] > 5]
)
```

```python
department_salary = (
    df.groupby("Department")["Salary"]
    .sum()
    .sort_values(ascending=False)
)

print(department_salary)
```

```python
department_salary = (
    df.groupby("Department")["Salary"]
    .sum()
    .sort_values(ascending=False)
)

print(department_salary)
```

```python
print(
    df.groupby("Department")["Age"].mean()
)
```

```python
print(
    df.sort_values("Salary", ascending=False)
)
```

```python
print(
    df.sort_values("Salary", ascending=False)
)
```
