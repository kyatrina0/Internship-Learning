```python
def add(a, b):
    return a + b

result = add(25, 15)

print("Sum =", result)
```

```python
def largest(a, b):
    if a > b:
        return a
    else:
        return b

print("Largest =", largest(18, 25))
```

```python
marks = []

for i in range(5):
    mark = int(input(f"Enter marks for Subject {i+1}: "))
    marks.append(mark)

total = sum(marks)
average = total / len(marks)

print("\nMarks:", marks)
print("Total:", total)
print("Average:", average)
```

```python
def is_prime(num):
    if num < 2:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True

number = int(input("Enter a number: "))

if is_prime(number):
    print(number, "is a Prime Number")
else:
    print(number, "is Not a Prime Number")
```
