```python
import numpy as np

data = np.random.normal(
    loc=70,
    scale=10,
    size=1000
)

print("Mean:", np.mean(data))
print("Standard Deviation:", np.std(data))
```

```python
import numpy as np

data = np.random.normal(
    loc=70,
    scale=10,
    size=1000
)

print("Mean:", np.mean(data))
print("Standard Deviation:", np.std(data))
```

```python
uniform_data = np.random.uniform(
    0,
    100,
    1000
)

print("Mean:", np.mean(uniform_data))
print("Standard Deviation:", np.std(uniform_data))
```

```python
normal_data = np.random.normal(50, 10, 1000)
uniform_data = np.random.uniform(0, 100, 1000)

print("Normal Mean:", np.mean(normal_data))
print("Uniform Mean:", np.mean(uniform_data))
```

```python
data = np.random.normal(70, 10, 10000)

between_60_80 = np.sum(
    (data >= 60) & (data <= 80)
)

percentage = between_60_80 / len(data) * 100

print("Percentage between 60 and 80:",
      percentage)
```
