```python
import pandas as pd
import numpy as np

df = pd.read_csv("student_performance.csv")

print(df)
```

```python
print("Average Score:", df["Average"].mean())
print("Median Score:", df["Average"].median())
print("Highest Score:", df["Average"].max())
print("Lowest Score:", df["Average"].min())
print("Standard Deviation:", df["Average"].std())
```

```python
total_students = len(df)
grade_a_students = len(df[df["Grade"] == "A"])

probability_a = grade_a_students / total_students

print("Grade A Probability:", probability_a)
print("Grade A Percentage:", probability_a * 100, "%")
```

```python
above_80 = len(df[df["Average"] > 80])

probability = above_80 / len(df)

print("Probability of score above 80:", probability)
```

```python
subject_averages = {
    "Math": df["Math"].mean(),
    "Science": df["Science"].mean(),
    "English": df["English"].mean()
}

print(subject_averages)

best_subject = max(
    subject_averages,
    key=subject_averages.get
)

print("Highest Average Subject:", best_subject)
```

```python
import matplotlib.pyplot as plt

plt.hist(df["Average"], bins=5)
plt.xlabel("Average Score")
plt.ylabel("Number of Students")
plt.title("Student Average Score Distribution")
plt.show()
```

```python
print("\n--- FINAL ANALYSIS ---")

print("Students:", len(df))
print("Average Score:", df["Average"].mean())
print("Median Score:", df["Average"].median())
print("Highest Score:", df["Average"].max())
print("Lowest Score:", df["Average"].min())
print("Standard Deviation:", df["Average"].std())
print("Best Subject:", best_subject)
print("Probability of Grade A:", probability_a)
```
