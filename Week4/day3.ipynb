```python
total_balls = 20
red_balls = 8
probability = red_balls / total_balls

print("Probability:", probability)
print("Percentage:", probability * 100, "%")
```

```python
import numpy as np

results = np.random.choice(
    ["Heads", "Tails"],
    size=100
)

heads = np.sum(results == "Heads")
tails = np.sum(results == "Tails")

print("Heads:", heads)
print("Tails:", tails)
print("Heads Probability:", heads / 100)
print("Tails Probability:", tails / 100)
```

```python
students = [
    "Pass", "Pass", "Fail", "Pass", "Fail",
    "Pass", "Pass", "Fail", "Pass", "Pass"
]

total = len(students)
passed = students.count("Pass")
failed = students.count("Fail")

print("Total:", total)
print("Passed:", passed)
print("Failed:", failed)

print("Probability of Pass:", passed / total)
print("Probability of Fail:", failed / total)
```

```python
numbers = np.random.randint(1, 101, 1000)

greater_than_50 = np.sum(numbers > 50)

probability = greater_than_50 / len(numbers)

print("Numbers greater than 50:", greater_than_50)
print("Probability:", probability)
```
