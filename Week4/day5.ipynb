```python
import pandas as pd
import numpy as np

data = {
    "Name": [
        "Alice", "Bob", "Charlie", "David",
        "Emma", "Frank", "Grace", "Henry",
        "Ivy", "Jack"
    ],
    "Math": [78, 85, 65, 90, 72, 88, 95, 60, 75, 82],
    "Science": [80, 82, 70, 92, 75, 85, 90, 65, 78, 88],
    "English": [75, 88, 72, 85, 80, 90, 92, 68, 76, 84]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Total"] = (
    df["Math"] +
    df["Science"] +
    df["English"]
)

print(df)
```

```python
df["Average"] = df["Total"] / 3

print(df)
```

```python
print(df.loc[df["Average"].idxmax()])
```

```python
print(df.loc[df["Average"].idxmin()])
```

```python
print("Math Average:", df["Math"].mean())
print("Science Average:", df["Science"].mean())
print("English Average:", df["English"].mean())
```

```python
print("Math Std:", df["Math"].std())
print("Science Std:", df["Science"].std())
print("English Std:", df["English"].std())
```

```python
overall_average = df["Average"].mean()

print(
    df[df["Average"] > overall_average]
)
```

```python
def grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 80:
        return "B"
    elif avg >= 70:
        return "C"
    elif avg >= 60:
        return "D"
    else:
        return "F"

df["Grade"] = df["Average"].apply(grade)

print(df)
```

```python
print(df["Grade"].value_counts())
```

```python
print(df[["Math", "Science", "English", "Total", "Average"]].describe())
df.to_csv("student_performance.csv", index=False)
```
