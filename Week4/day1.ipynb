```python
import numpy as np
import pandas as pd
marks = [65, 70, 75, 80, 80, 85, 90]
mean = np.mean(marks)
print("Mean:", mean)
```

```python
median = np.median(marks)
print("Median:", median)
```

```python
from statistics import mode

print("Mode:", mode(marks))
```

```python
marks = np.array([65, 70, 75, 80, 80, 85, 90])

data_range = np.max(marks) - np.min(marks)

print("Range:", data_range)
```

```python
marks = np.array([65, 70, 75, 80, 80, 85, 90])

variance = np.var(marks)
std = np.std(marks)

print("Variance:", variance)
print("Standard Deviation:", std)
```

```python
marks = np.array([65, 70, 75, 80, 80, 85, 90])

print("25th percentile:", np.percentile(marks, 25))
print("50th percentile:", np.percentile(marks, 50))
print("75th percentile:", np.percentile(marks, 75))
```

```python
students = 10
passed = 7
probability = passed / students

print("Probability:", probability)
```

```python
results = ["Pass", "Pass", "Fail", "Pass", "Fail",
           "Pass", "Pass", "Pass", "Fail", "Pass"]

total = len(results)
passed = results.count("Pass")

probability = passed / total

print("Total students:", total)
print("Passed students:", passed)
print("Probability of passing:", probability)
```
