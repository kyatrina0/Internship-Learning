# WEEK 9 - TIME SERIES BASICS

Retail Sales Analysis

## 1. Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Load dataset

```python
data = pd.read_csv("sales_time_series.csv")

print("First 5 rows:")
print(data.head())

print("\nDataset shape:")
print(data.shape)

print("\nColumn names:")
print(data.columns)
```

## 3. Basic information

```python
print("\nDataset information:")
data.info()

print("\nMissing values:")
print(data.isnull().sum())

print("\nBasic statistics:")
print(data.describe())
```

## 4. Check the column names

```python
print("\nColumn names in the dataset:")
print(data.columns.tolist())
```

## 5. Convert date column to datetime

```python
# Change these names if your dataset has different column names

if "date" in data.columns:
    date_column = "date"
elif "Date" in data.columns:
    date_column = "Date"
else:
    date_column = data.columns[0]

if "sales" in data.columns:
    sales_column = "sales"
elif "Sales" in data.columns:
    sales_column = "Sales"
else:
    sales_column = data.select_dtypes(include="number").columns[0]

print("\nDate column:", date_column)
print("Sales column:", sales_column)


data[date_column] = pd.to_datetime(data[date_column], errors="coerce")

# Remove rows with missing date or sales

data = data.dropna(subset=[date_column, sales_column])

# Sort by date

data = data.sort_values(date_column)

print("\nDate range:")
print(data[date_column].min(), "to", data[date_column].max())
```

## 6. Set date as index

```python
data = data.set_index(date_column)

print("\nData after setting date as index:")
print(data.head())
```

## 7. Sales over time

```python
plt.figure(figsize=(10, 5))

plt.plot(data.index, data[sales_column])

plt.title("Sales Over Time")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 8. Daily sales

```python
daily_sales = data[sales_column].resample("D").sum()

print("\nDaily sales:")
print(daily_sales.head(10))
```

## 9. Weekly sales

```python
weekly_sales = data[sales_column].resample("W").sum()

print("\nWeekly sales:")
print(weekly_sales.head(10))
```

## 10. Monthly sales

```python
monthly_sales = data[sales_column].resample("ME").sum()

print("\nMonthly sales:")
print(monthly_sales)
```

## 11. Daily sales graph

```python
plt.figure(figsize=(10, 5))

plt.plot(daily_sales.index, daily_sales.values)

plt.title("Daily Sales")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 12. Weekly sales graph

```python
plt.figure(figsize=(10, 5))

plt.plot(weekly_sales.index, weekly_sales.values)

plt.title("Weekly Sales")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 13. Monthly sales graph

```python
plt.figure(figsize=(10, 5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 14. 7 day rolling average

```python
rolling_sales = daily_sales.rolling(7).mean()

print("\n7 day rolling average:")
print(rolling_sales.head(15))
```

## 15. Daily sales and rolling average

```python
plt.figure(figsize=(10, 5))

plt.plot(
    daily_sales.index,
    daily_sales.values,
    label="Daily Sales"
)

plt.plot(
    rolling_sales.index,
    rolling_sales.values,
    label="7 Day Rolling Average"
)

plt.title("Daily Sales and 7 Day Rolling Average")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.legend()

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 16. Find highest and lowest sales

```python
highest_sales = daily_sales.max()
lowest_sales = daily_sales.min()

highest_date = daily_sales.idxmax()
lowest_date = daily_sales.idxmin()

print("\nHighest sales:")
print(highest_sales)

print("Date of highest sales:")
print(highest_date)

print("\nLowest sales:")
print(lowest_sales)

print("Date of lowest sales:")
print(lowest_date)
```

## 17. Simple moving average forecast

```python
last_7_days = daily_sales.tail(7)

forecast = last_7_days.mean()

print("\nAverage sales of the last 7 days:")
print(round(forecast, 2))
```

## 18. Forecast next 7 days

```python
future_dates = pd.date_range(
    start=daily_sales.index[-1] + pd.Timedelta(days=1),
    periods=7
)

forecast_values = [forecast] * 7

forecast_data = pd.DataFrame({
    "Date": future_dates,
    "Forecast Sales": forecast_values
})

print("\nForecast for next 7 days:")
print(forecast_data)
```

## 19. Forecast graph

```python
plt.figure(figsize=(10, 5))

recent_sales = daily_sales.tail(30)

plt.plot(
    recent_sales.index,
    recent_sales.values,
    label="Actual Sales"
)

plt.plot(
    future_dates,
    forecast_values,
    marker="o",
    label="Forecast"
)

plt.title("Simple Sales Forecast")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.legend()

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 20. Compare first and last part of the data

```python
first_sales = daily_sales.head(7).mean()
last_sales = daily_sales.tail(7).mean()

print("\nAverage sales of first 7 days:")
print(round(first_sales, 2))

print("\nAverage sales of last 7 days:")
print(round(last_sales, 2))

if last_sales > first_sales:
    print("\nThe recent sales are higher than the beginning of the dataset.")
elif last_sales < first_sales:
    print("\nThe recent sales are lower than the beginning of the dataset.")
else:
    print("\nThe recent sales are almost similar to the beginning of the dataset.")
```

## 21. Average sales by day of week

```python
day_sales = daily_sales.groupby(
    daily_sales.index.day_name()
).mean()

day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_sales = day_sales.reindex(day_order)

print("\nAverage sales by day:")
print(day_sales)
```

## 22. Day of week graph

```python
plt.figure(figsize=(10, 5))

plt.bar(
    day_sales.index,
    day_sales.values
)

plt.title("Average Sales by Day of Week")
plt.xlabel("Day")
plt.ylabel("Average Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 23. Final statistics

```python
print("\n" + "=" * 50)
print("FINAL SUMMARY")
print("=" * 50)

print("Average daily sales:", round(daily_sales.mean(), 2))
print("Highest daily sales:", round(daily_sales.max(), 2))
print("Lowest daily sales:", round(daily_sales.min(), 2))
print("Average weekly sales:", round(weekly_sales.mean(), 2))
print("Average monthly sales:", round(monthly_sales.mean(), 2))
```
